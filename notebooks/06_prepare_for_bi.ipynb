```python
import pandas as pd

master = pd.read_csv(
    '../data/processed_data/master_orders.csv',
    parse_dates=['order_purchase_timestamp']
)

# Для каждого клиента — вернулся ли он
customer_orders = master.groupby('customer_unique_id').size()
master['is_returning'] = master['customer_unique_id'].map(customer_orders >= 2)

# Месяц в формате YYYY-MM (для осей графиков)
master['order_month_str'] = master['order_purchase_timestamp'].dt.strftime('%Y-%m')

# Год
master['order_year'] = master['order_purchase_timestamp'].dt.year

# Категория оценки (для фильтров)
master['review_category'] = pd.cut(
    master['review_score'],
    bins=[0, 2, 3, 4, 5],
    labels=['Плохо (1-2)', 'Средне (3)', 'Хорошо (4)', 'Отлично (5)']
)

# Сохраняем для Power BI
master.to_csv(
    '../data/processed_data/master_for_bi.csv',
    index=False,
    sep=';',              # разделитель колонок — точка с запятой
    decimal=',',          # десятичный разделитель — запятая
    encoding='utf-8-sig'  # BOM для корректной работы с русскими символами
)

print("Сохранено: master_for_bi.csv")
print(f"Размер: {len(master):,} строк, {len(master.columns)} колонок")
print()
print("Новые колонки:")
print(master[['is_returning', 'order_month_str', 'order_year', 'review_category']].head())
```
